```python
def within_tissue_batch_silhouette(data, metadata, batch_key, tissue_key="SMTSD", n_pcs=30, min_samples=50):
    from sklearn.decomposition import PCA
    from sklearn.metrics import silhouette_score
    import numpy as np

    scores = {}
    weights = {}

    for tissue in metadata[tissue_key].unique():
        tissue_mask = (metadata[tissue_key] == tissue).values

        # Drop samples with NaN in the batch key
        valid_mask = tissue_mask & metadata[batch_key].notna().values

        n_samples = valid_mask.sum()
        n_batches = metadata.loc[valid_mask, batch_key].nunique()

        if n_samples < min_samples or n_batches < 2:
            continue

        tissue_data = data.loc[:, valid_mask]
        pca = PCA(n_components=min(n_pcs, n_samples - 1))
        pcs = pca.fit_transform(tissue_data.T)

        batch_labels = metadata.loc[valid_mask, batch_key].values
        score = silhouette_score(pcs, batch_labels)

        scores[tissue] = score
        weights[tissue] = n_samples

    total = sum(weights.values())
    weighted_score = sum(scores[t] * weights[t] / total for t in scores)

    return weighted_score, scores

# Compare batch variables properly
geo_score, geo_per_tissue = within_tissue_batch_silhouette(df, meta_filtered, "SUBJID")
# na_score,  na_per_tissue  = within_tissue_batch_silhouette(df, meta_filtered, "SMNABTCH")
# sm_score, sm_per_tissue = within_tissue_batch_silhouette(df, meta_filtered, "SMCENTER")

print(f"SMGEBTCH: {geo_score:.3f}")
# print(f"SMNABTCH: {na_score:.3f}")
# print(f"SMCENTER:   {sm_score:.3f}")
```
