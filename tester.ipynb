```python
import pandas as pd
import matplotlib.pyplot as plt
import harmonypy as hm
from sklearn.decomposition import PCA
import umap

df = pd.read_csv("Dataset/result/ML_Dataset/expression_dataframe.csv", index_col=0)
df.fillna(0, inplace=True)
meta2 = pd.read_csv("Dataset/drive-download/metadata/GTEx_Analysis_v10_Annotations_SampleAttributesDS.txt", sep="\t")


meta_filtered = meta2.loc[meta2["SAMPID"].isin(df.columns), ["SAMPID", "SMGEBTCH", "SMTSD"]]
meta_filtered["SUBJID"] = meta_filtered["SAMPID"].str.split("-").str[:2].str.join("-")
meta_filtered = meta_filtered.set_index("SAMPID").reindex(df.columns)
meta_filtered.index.name = "SAMPID"
meta_filtered = meta_filtered.reset_index()

meta_filtered
```

```python
meta_filtered
```

```python
meta_filtered["SAMPID"]
```

```python
print(all(meta_filtered["SAMPID"] == df.columns.tolist()))  # should be True
```

```python

pca = PCA(n_components=30)
pcs = pca.fit_transform(df.T)
pcs = pd.DataFrame(pcs, index=df.columns)


harmony_out = hm.run_harmony(pcs, meta_filtered, ["SUBJID", "SMGEBTCH"])

harmony_result = pd.DataFrame(
    harmony_out.Z_corr,
    index=pcs.index
)

reducer = umap.UMAP(random_state=42)

embedding_before = umap.UMAP(random_state=42).fit_transform(pcs)
embedding_after = umap.UMAP(random_state=42).fit_transform(harmony_result)
```

```python
donor_cat = meta_filtered["SUBJID"].astype("category")
donor_codes = donor_cat.cat.codes
donor_labels = donor_cat.cat.categories

handles = [plt.Line2D([0],[0], marker='o', color='w',
            markerfacecolor=plt.cm.tab20(i/len(donor_labels)),
            label=label) for i, label in enumerate(donor_labels)]
axes[1].legend(handles=handles, fontsize=6, bbox_to_anchor=(1.05, 1))


fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# BEFORE
axes[0].scatter(
    embedding_before[:, 0],
    embedding_before[:, 1],
    c=donor_codes,
    s=5,
    cmap="tab20"
)
axes[0].set_title("Before Harmony")
axes[0].set_xlabel("UMAP1")
axes[0].set_ylabel("UMAP2")

# AFTER
axes[1].scatter(
    embedding_after[:, 0],
    embedding_after[:, 1],
    c=donor_codes,
    s=5,
    cmap="tab20"
)
axes[1].set_title("After Harmony")
axes[1].set_xlabel("UMAP1")
axes[1].set_ylabel("UMAP2")


plt.tight_layout()
plt.show()
```
